**Pyhton Function**

A python function is a reuseable block of code designed to perform a specific task.

```python
def calculate_total(price, qty):
  total=price*qty
  return total

  bill= calculate_total(120, 3)
  print("total:", bill)
```

Functions are useful because they avoid repeating code, make programs easier to organize, and let you break a large problem into smaller pieces.

**Basic structure of a python function**


```python
def function_name(parameter1, parameter2):
  # indented function body
  result = parameter1 + parameter2
  return result
```

**Parameters vs Arguments**

```python
def add (a,b):
  return(a+b)

answer = add(10,5)
print(answer)
```

In the above code, (a,b) are parameters, while (10,5) are arguments.

**The return statement**
- return sends a value back from the function. That value can be stored, printed or reused.

```python
def square (n):
  return n*n

x=square (6)
print(x)
print(square(4))
```

**Print vs Return**
- when to use it?

  **print()**
- shows text on the screen
- good for output messages and quick debugging
- does not give a reuseable value back
- use print near the user interface

  **return**
- sends a value back to the caller
- good for calculation, logic, testing and reuse
- The value can be stored in a variable
- use return inside logic functions

```python
def add_print(a,b):
  print(a+b)

x=add_print(2,3)
print(x)
```

**Mini Program**
- celsius to fahrenheit

```python
def celsius_to_fahrenheit(celsius):
  fahrenheit=(celsius*9/5)+32
  return fahrenheit

  temp_c = float(input("Enter the celsius:"))
  temp_f = celsius_to_fahrenheit(temp_c)
print ("fahrenheit:", temp_f)

```

**Function call flow: from input to output**

1. Input  (a=20, b=5)
2. call [divide(a,b)]
3. Process (a/b)
4. Return (4.0)
5. Print [Result:4.0]

```python
def divide (a,b):
  return a/b

result = divide(20,5)
print("Result:", result)
```
