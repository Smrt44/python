**python Tuple**

- A tuple stores multiple values in one variable.

```python
student = ("Ram", 23, "BBS")
marks = (78, 67, 89)
print(student[0])
print(student[2])
```

**Tuple characterstics**

- Ordered: items keep their positions
- indexed: each value has an index number
- immutable: values cannot be changed after creation
- allows duplicate values
- can store mixed data types.

```python
person=("Ram", 25, "Kathmandu")
print(person)
print(type(person))
```

**Tuple Slicing**

- Use slicing when you need only a selected range of values.

```python
marks = (26, 56, 78, 89, 99, 200)
print(marks[1:4])
print(marks[:4])
print(marks[::3])
print(marks[-1:-4])
```

**Tuple with mutable items**
- Tuple cannot replace items, but if the item is a list, the list content can be changed.

```python
data=([10,20], "ram")
data[0].append("hari")
print(data)
data[1].append("shyam")
```
