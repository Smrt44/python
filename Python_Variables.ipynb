**Python** **Variables**

Names that store values in a program.

Examples: age=21, name="Samir", marks=90.5, is_passed=true

**Main Rules:**
- Start with name or underscore
- cannot start with a number
- use only letters, digits and underscore
- no spaces
- no special symbools
- Upper/Lower case sensetive:(Age,age)
- cannot use python keywords:if, class, for, true, non
- use meaningful names: marks>m
- avoid built-in names:list, str, sum, input

```python
# create variables
student_name="Sam"
age=27
marks=90.5
is_passed=marks>=40

#update variable value
age=age+3
print(student_name, age)

```

**Updating Variable values**

A variable can point to a new value later

```python
score=56
score=score+10
score+=4
print(score)
```

**Multiple Assignment**

Python can assign values to several variables in one line.

Useful for:

- clean code
- swapping values
- unpacking data
- initializing many variables

```python
a, b, c= 10, 20, 30
print(a,b,c)
```

```python
x=y=z=10
print(x, y, z)
first_name, last_name="Hari", "Khadka"
full_name= first_name +" "+ last_name
print(full_name)
```
